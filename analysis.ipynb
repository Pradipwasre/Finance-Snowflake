# Finance ML Dataset - Quick Analysis

This notebook loads the data pulled from Snowflake (`connection.py`) and does a
quick pandas-based exploration.

**Two ways to load data here:**
1. Re-run the Snowflake connector directly in this notebook (live pull)
2. Load the cached local CSV created by `connection.py` (faster, no warehouse cost)

Use Option 2 for repeated practice; use Option 1 to show students the live connection.

## Option 1: Load data directly from Snowflake (live)

```python
import pandas as pd
import snowflake.connector

conn = snowflake.connector.connect(
    connection_name="my_example_connection"
)

query = """
    SELECT *
    FROM FINANCE_ML_DB.FINANCE_PROJECTS.FINANCE_PROJECTS_ML_DATA
"""

df = pd.read_sql(query, conn)
conn.close()

df.shape
```

## 3. Group-level business insight

```python
df.groupby("FINANCE_DOMAIN")["ACTUAL_PROJECT_COST"].mean().sort_values(ascending=False)
```

```python
df.groupby("PROJECT_TYPE")["ACTUAL_PROJECT_COST"].agg(["mean", "count"]).sort_values("mean", ascending=False)
```

## 4. Correlation with the ML target (ACTUAL_PROJECT_COST)

```python
numeric_df = df.select_dtypes(include="number")
numeric_df.corr()["ACTUAL_PROJECT_COST"].sort_values(ascending=False)
```

## 5. Quick visualization

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["ACTUAL_PROJECT_COST"], bins=40)
plt.title("Distribution of Actual Project Cost")
plt.xlabel("Actual Project Cost")
plt.ylabel("Frequency")
plt.show()
```
